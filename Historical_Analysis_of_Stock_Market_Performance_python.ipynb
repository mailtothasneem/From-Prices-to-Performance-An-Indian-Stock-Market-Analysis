## Hypothesis Testing(T-Test)

## Independent t-test: RELIANCE vs WIPRO

- The t-statistic is 0.916 and the p-value is 0.360.

- Since the p-value is greater than 0.05, this test does not provide enough evidence to conclude that RELIANCE and WIPRO had different mean daily returns in this dataset.

- This does not prove their mean returns were exactly the same.

- It means the observed difference was not statistically significant at the 5% level.

```python
from scipy.stats import ttest_ind
returns1 = df.loc[df["Symbol"] == "RELIANCE", "Daily_Return"].dropna()
returns2 = df.loc[df["Symbol"] == "WIPRO", "Daily_Return"].dropna()



t_stat, p_value = ttest_ind(returns1, returns2, equal_var=False)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

## ANOVA

### One-Way ANOVA of Daily Returns

- A one-way ANOVA was used to compare the mean daily returns across the ticker groups.

- The F-statistic was 0.694 and the p-value was 0.628.

- Since the p-value is greater than 0.05, there is not enough evidence to conclude that the tickers have different mean daily returns in this dataset. This does not prove that their means are exactly equal.

```python
from scipy.stats import f_oneway

# Get daily returns for each ticker

return_groups = [group["Daily_Return"].dropna().values
    for _, group in df.groupby("Symbol")
]


f_stat, p_value = f_oneway(*return_groups)

print("F-statistic:", f_stat)
print("p-value:", p_value)
```

## Chi- Square Test

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Creating return categories

df["Return_Category"] = df["Daily_Return"].apply(
    lambda x: "Positive" if x > 0 else "Zero_or_Negative"
    if pd.notna(x) else None
)

# Count each category for each ticker
table = pd.crosstab(df["Symbol"], df["Return_Category"])

display(table)

# H0: return category is independent of ticker
# H1: return category is associated with ticker

chi2_stat, p_value, degrees_freedom, expected = chi2_contingency(table)

print("Chi-square statistic:", chi2_stat)
print("Degrees of freedom:", degrees_freedom)
print("p-value:", p_value)
```
